```python
import pandas as pd
seed_files = {
    'seed_1perc': '/content/drive/MyDrive/THESIS/05outputs/dev_seed_1percent.csv',
    'seed_5perc': '/content/drive/MyDrive/THESIS/05outputs/dev_seed_5percent.csv',
    'seed_10perc': '/content/drive/MyDrive/THESIS/05outputs/dev_seed_10percent.csv',
    'seed_20perc': '/content/drive/MyDrive/THESIS/05outputs/dev_seed_20percent.csv'
}

seed_dfs = {}
for name, path in seed_files.items():
    try:
        seed_dfs[name] = pd.read_csv(path)
        print(f"Loaded {name}: {len(seed_dfs[name])} rows")
    except FileNotFoundError:
        print(f"Warning: Seed file not found at {path}")

print("Seed files loaded.")
```

### Applying SMOTE (Synthetic Minority Over-sampling Technique)

SMOTE is an oversampling technique used to address imbalanced datasets by generating synthetic samples from the minority class. This helps in training machine learning models more effectively when one class is significantly underrepresented.

```python
# Install imbalanced-learn if not already installed
!pip install imbalanced-learn
```

```python
import pandas as pd
from collections import Counter
from imblearn.over_sampling import SMOTE

TARGET_COLUMN_NAME = 'Churn Value'

if TARGET_COLUMN_NAME in seed_dfs['seed_1perc'].columns:
    X = seed_dfs['seed_1perc'].drop(TARGET_COLUMN_NAME, axis=1)
    y = seed_dfs['seed_1perc'][TARGET_COLUMN_NAME]

    print(f"Original dataset shape: {Counter(y)}")

    # Modify SMOTE to generate approximately 5000 rows (2500 per class)
    # The sampling_strategy dictionary specifies the desired number of samples for each class after resampling.
    sm = SMOTE(sampling_strategy={0: 2500, 1: 2500}, random_state=42)
    X_res, y_res = sm.fit_resample(X, y)

    print(f"Resampled dataset shape: {Counter(y_res)}")

    # Create a new DataFrame with the SMOTE-resampled data
    seed_1perc_smoted = pd.DataFrame(X_res, columns=X.columns)
    seed_1perc_smoted[TARGET_COLUMN_NAME] = y_res

    print("\nFirst 5 rows of the SMOTE-resampled data:")
    display(seed_1perc_smoted.head())
    print(f"Total rows in resampled DataFrame: {len(seed_1perc_smoted)}")
else:
    print(f"Error: The column '{TARGET_COLUMN_NAME}' was not found in the DataFrame. Please check the column name.")
```

The `seed_1perc` DataFrame has now been oversampled using SMOTE, and the class distribution of the 'Churn' variable should be balanced.

```python
output_path = '/content/drive/MyDrive/THESIS/05outputs/resultSMOTE1p.csv'
seed_1perc_smoted.to_csv(output_path, index=False)
print(f"SMOTEd data saved to: {output_path}")
```
